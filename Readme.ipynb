```python
Magnetic Field Energy Form (D8)
B.Tech Mathematics Mini Project
A mathematical and computational study of the magnetic energy stored in three magnetically coupled coils using quadratic forms, positive definiteness, eigenvalues, and principal inductances.


1. Project Overview
The Magnetic Field Energy Form (D8) project connects concepts from linear algebra and quadratic forms with a real-world electrical engineering application.

For three coupled coils, the magnetic energy stored in the system can be represented as:

[
W = \frac{1}{2}\mathbf{i}^{T}L\mathbf{i}
]

where:


W = stored magnetic energy in joules
\mathbf{i} = current vector
L = inductance matrix
L^T=L because the inductance matrix is symmetric

The project analyzes this quadratic form and determines whether the energy form is positive definite, positive semidefinite, or indefinite.


2. Problem Statement

Three coils have self-inductances and mutual inductances. The energy stored is given by


W=\frac{1}{2}\mathbf{i}^{T}L\mathbf{i}


where L is a symmetric inductance matrix. Classify the energy quadratic form. Is it always positive definite? What does that mean physically? Find the principal inductances.



3. Mathematical Model
For three coupled coils, the inductance matrix is:

[
L=
\begin{bmatrix}
L_1 & M_{12} & M_{13}\
M_{12} & L_2 & M_{23}\
M_{13} & M_{23} & L_3
\end{bmatrix}
]

and the current vector is:

[
\mathbf{i}=
\begin{bmatrix}
i_1\
i_2\
i_3
\end{bmatrix}
]

Therefore,

[
W=\frac{1}{2}\mathbf{i}^{T}L\mathbf{i}
]

Expanding the quadratic form:

[
W=\frac{1}{2}
\left[
L_1i_1^2+
L_2i_2^2+
L_3i_3^2+
2M_{12}i_1i_2+
2M_{13}i_1i_3+
2M_{23}i_2i_3
\right]
]


4. Main Mathematical Concepts
4.1 Quadratic Form
A quadratic form is an expression of the type:

[
Q(\mathbf{x})=\mathbf{x}^{T}A\mathbf{x}
]

In this project:

[
Q(\mathbf{i})=\mathbf{i}^{T}L\mathbf{i}
]

and the magnetic energy is:

[
W=\frac{1}{2}Q(\mathbf{i})
]


4.2 Positive Definiteness
For a symmetric matrix L, the quadratic form is positive definite if:

[
\mathbf{i}^{T}L\mathbf{i}>0
]

for every non-zero current vector \mathbf{i}.

In physical terms, this means that every non-zero current state stores positive magnetic energy.


4.3 Eigenvalues
The eigenvalues of L are obtained from:

[
\det(L-\lambda I)=0
]

For a symmetric inductance matrix, the eigenvalues are real.

The eigenvalues are used to determine the definiteness of the quadratic form.


4.4 Principal Inductances
The eigenvalues of the inductance matrix are called the principal inductances:

[
\lambda_1,\lambda_2,\lambda_3
]

They represent the inductive behavior of the system along its principal modes.


5. Physical Interpretation
For a passive magnetic system, stored energy cannot be negative.

Therefore, physically meaningful inductance parameters should make the energy form at least positive semidefinite:

[
W\geq0
]

If the matrix is positive definite:

[
W>0
]

for every non-zero current vector.

This means the system stores strictly positive magnetic energy whenever current is present.


6. Is the Energy Form Always Positive Definite?
No.

Positive definiteness depends on the values of the self-inductances and mutual inductances.

For example, excessively large or inconsistent mutual inductances can make the inductance matrix fail the positive-definiteness condition.

For a physically valid passive coupled-coil system, the inductance matrix must satisfy the appropriate physical constraints so that the stored energy is never negative.

The program automatically checks the eigenvalues and classifies the quadratic form.


7. Positive Definiteness Using Eigenvalues
For the symmetric matrix L:

Positive Definite
If:

[
\lambda_1>0,\quad
\lambda_2>0,\quad
\lambda_3>0
]

then L is positive definite.

Positive Semidefinite
If:

[
\lambda_k\geq0
]

for all k, with at least one zero eigenvalue, then L is positive semidefinite.

Indefinite
If some eigenvalues are positive and some are negative, the quadratic form is indefinite.

The project uses eigenvalues to perform this classification automatically.


8. Principal Minors
The project also calculates the leading principal minors:

[
D_1=L_1
]

[
D_2=
\begin{vmatrix}
L_1&M_{12}\
M_{12}&L_2
\end{vmatrix}
]

and

[
D_3=\det(L)
]

For a real symmetric matrix, Sylvester's criterion states that the matrix is positive definite when all leading principal minors are positive:

[
D_1>0,\qquad D_2>0,\qquad D_3>0
]

The project therefore provides both eigenvalue-based classification and principal-minor analysis.


9. Computational Workflow
The project follows these steps:


Accept the self-inductances from the user.
Accept the mutual inductances from the user.
Accept the three coil currents.
Construct the symmetric inductance matrix.
Construct the magnetic-energy quadratic form.
Calculate the stored magnetic energy.
Calculate the eigenvalues and eigenvectors.
Identify the principal inductances.
Classify the quadratic form.
Calculate the principal minors.
Interpret the mathematical result physically.
Generate visual graphs.


10. User Input
The program does not use fixed values only.

It dynamically asks the user to enter:

Self-Inductances

L_1
L_2
L_3

Mutual Inductances

M_{12}
M_{13}
M_{23}

Currents

i_1
i_2
i_3

The entered values are then used to calculate the complete mathematical solution.


11. Example Input
A sample input is:

L1  = 4 H
L2  = 3 H
L3  = 2.5 H

M12 = 0.8 H
M13 = 0.5 H
M23 = 0.6 H

i1 = 2 A
i2 = 1.5 A
i3 = 1 A

The corresponding inductance matrix is:

[
L=
\begin{bmatrix}
4&0.8&0.5\
0.8&3&0.6\
0.5&0.6&2.5
\end{bmatrix}
]

For these values, the stored magnetic energy is:

[
\boxed{W=13.45\text{ J}}
]

The program also calculates the eigenvalues and classifies the matrix as positive definite.


12. Visualizations
The project generates three main visualizations.

1. Principal Inductance Chart
A bar chart showing the three principal inductances obtained from the eigenvalues.

2. Magnetic Energy Curve
A graph showing how magnetic energy changes with the magnitude of a selected current direction.

For a positive-definite energy form, the graph has an upward-opening quadratic shape.

3. 3D Energy Surface
A three-dimensional visualization of the quadratic energy function for a selected current plane.

These visualizations help connect the mathematical quadratic form with its physical interpretation.


13. Technologies Used

Python
NumPy — numerical matrix calculations and eigenvalue computation
SymPy — symbolic mathematical expressions
Matplotlib — graphs and visualizations
Google Colab — development and execution environment


14. Project Structure
Magnetic-Field-Energy-Form-D8/
│
├── README.md
│
└── Magnetic_Field_Energy_Form.ipynb

The notebook contains:

1. Import Libraries
2. User Input
3. Inductance Matrix Construction
4. Quadratic Form
5. Magnetic Energy Calculation
6. Eigenvalue Analysis
7. Positive Definiteness Test
8. Principal Inductances
9. Principal Minors
10. Physical Interpretation
11. Data Visualization


15. Real-World Applications
The mathematical model is related to several electrical engineering applications, including:


Transformers
Electric motors
Magnetically coupled coils
Inductors
Electrical energy-storage systems
Coupled electromagnetic circuits

The eigenvalue analysis provides a mathematical way to understand the independent principal modes of the coupled system.


16. Learning Outcomes
Through this project, we learn how to:


Represent a physical system using matrices.
Formulate magnetic energy as a quadratic form.
Understand positive definiteness.
Calculate eigenvalues and eigenvectors.
Interpret eigenvalues as principal inductances.
Connect linear algebra with electrical engineering.
Use Python for mathematical computation.
Visualize mathematical functions using graphs.
Understand the physical meaning of mathematical results.


17. Assumptions
The following assumptions are made:


The system consists of three magnetically coupled coils.
The inductance matrix is symmetric.
Self-inductances and mutual inductances are represented in henries.
Currents are represented in amperes.
The standard magnetic-energy relation is used:

[
W=\frac12\mathbf{i}^{T}L\mathbf{i}
]


The analysis assumes a passive inductive system when interpreting physical validity.
Numerical eigenvalue calculations use standard linear algebra methods.


18. Conclusion
This project demonstrates how a real-world magnetic energy problem can be represented using a quadratic form.

The inductance matrix contains the self-inductances and mutual inductances of the three coils. By calculating its eigenvalues, we can determine the principal inductances and classify the energy form.

The positive-definiteness analysis provides an important physical interpretation: a valid passive magnetic system should not produce negative stored energy.

Thus, the project successfully connects:

[
\boxed{\text{Linear Algebra}}
]

with

[
\boxed{\text{Quadratic Forms}}
]

and

[
\boxed{\text{Electrical Engineering}}
]

through the mathematical model:

[
\boxed{W=\frac12\mathbf{i}^{T}L\mathbf{i}}
]


19. Future Improvements
Possible future extensions include:


Allowing any number of coupled coils.
Adding interactive input controls.
Adding contour plots of the energy surface.
Comparing different coupling conditions.
Studying how mutual inductance affects eigenvalues.
Adding a graphical user interface.
Exporting calculated results automatically.
Comparing theoretical and experimental measurements.


20. Project Keywords
Quadratic Form

Positive Definiteness

Eigenvalues

Eigenvectors

Principal Inductances

Magnetic Energy

Mutual Inductance

Inductance Matrix

Linear Algebra

Electrical Engineering

Python

NumPy

SymPy

Matplotlib

Google Colab
```
